```python
import matplotlib.pyplot as plt
import mplhep as hep
import pandas as pd
import numpy as np
import config
hep.style.use(hep.style.CMS)

df_Sol150_initial = pd.read_csv("Sol150/PiPlus/initial_det_counts.csv")
df_Sol150_pcut = pd.read_csv("Sol150/PiPlus/pcut_det_counts.csv")
df_TrgCenter_initial = pd.read_csv("TrgCenter/PiPlus/initial_det_counts.csv")
df_TrgCenter_pcut = pd.read_csv("TrgCenter/PiPlus/pcut_det_counts.csv")
```

```python
from IPython.display import display, HTML
display(HTML(df_Sol150_initial.to_html(index=False)))
```

```python
plt.figure(figsize=(8, 8))

for _, row in df_Sol150_initial.iterrows():
    Bfield = f"B{row['B (T)']:02d}"
    plt.plot(
        stages,
        row[stages],
        marker="o",
        label=f"{row['B (T)']} T",
        color=config.COLOR_MAP[Bfield]
    )

plt.xlabel("Stage", fontsize=16)
plt.ylabel(r"$\pi^+$ count", fontsize=16)
plt.ylim(0,1000000)
plt.legend(fontsize=14)
plt.tick_params(axis='both', labelsize=16)
plt.title(fr"Cutflow of $\pi^+$ in 1.5 m solenoid [initial]", fontsize=18)
plt.grid(alpha=0.5)
plt.tight_layout()
plt.show()
```
